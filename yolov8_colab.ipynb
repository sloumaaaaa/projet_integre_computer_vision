```python
# 4. Train YOLOv8 on your dataset
from ultralytics import YOLO

# Path to your data.yaml (update if needed)
data_yaml = '/content/drive/MyDrive/your_project/data.yaml'  # <-- CHANGE THIS

# Start training
model = YOLO('yolov8n.pt')  # You can use yolov8s.pt, yolov8m.pt, etc.
results = model.train(
    data=data_yaml,
    epochs=50,
    imgsz=640,
    project='output',
    name='yolov8_exp',
    exist_ok=True
)
```

```python
# 5. Evaluate the trained model
metrics = model.val(data=data_yaml)
print('Evaluation metrics:', metrics)
```

```python
# 6. Test on a single image
test_img = '/content/drive/MyDrive/your_project/test_image.jpg'  # <-- CHANGE THIS
results = model(test_img)
results.show()  # Show image with detections
```
